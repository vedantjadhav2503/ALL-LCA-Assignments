```python
# Student Data Management using Dictionary, Tuple, and List

# Dictionary: roll number -> (name, branch, marks_list)
students = {
    101: ("Samey", "CSE", [85, 90, 88]),
    102: ("Anish", "CSE", [78, 82, 80]),
    103: ("Rajat", "ME", [72, 75, 70])
}

# 1. Add a new student record
students[104] = ("Arnav", "IT", [91, 89, 94])

# 2. Delete an existing student record
del students[103]

# 3. Update a student's details
students[102] = ("Anish Jadhav", "CSE", [84, 86, 88])

# 4. Display final records
print("Final Student Records:")
for roll_no, details in students.items():
    name, branch, marks = details
    print(f"\nRoll Number: {roll_no}")
    print(f"Name: {name}")
    print(f"Branch: {branch}")
    print(f"Marks: {marks}")
    print(f"Average Marks: {sum(marks) / len(marks):.2f}")
```
